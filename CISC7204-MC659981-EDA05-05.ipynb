```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('5. London Housing Data.csv')

print(df.head())
print(df.info())
```

```python
df['date'] = pd.to_datetime(df['date'])

print(df['date'].dtype)
print(df.head())
```

```python
df['year'] = df['date'].dt.year

month_data = df['date'].dt.month
df.insert(loc=1, column='month', value=month_data)

print(df.head())
```

```python
df.drop(['year', 'month'], axis=1, inplace=True)

print(df.columns)
print(df.head())
```

```python
zero_crimes_records = df[df['no_of_crimes'] == 0]

print("Records where 'no_of_crimes' is 0:")
print(zero_crimes_records)

count_zero_crimes = len(zero_crimes_records)
print(f"\nTotal number of records with 0 crimes: {count_zero_crimes}")
```

```python
df['year'] = df['date'].dt.year

england_df = df[df['area'] == 'england']

england_price_per_year = england_df.groupby('year')['average_price'].agg(['max', 'min'])

print("Maximum and Minimum average price per year in England:")
print(england_price_per_year)

df.drop('year', axis=1, inplace=True)
```

```python
crimes_per_area = df.groupby('area')['no_of_crimes'].agg(['max', 'min'])

print("Maximum and Minimum number of crimes recorded per area:")
print(crimes_per_area)
```

```python
low_price_df = df[df['average_price'] < 100000]

area_counts_low_price = low_price_df['area'].value_counts()

print("Total count of records for each area where average price is less than 100000:")
print(area_counts_low_price)
```
