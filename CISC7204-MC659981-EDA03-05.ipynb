```python
import pandas as pd
```

```python
df = pd.read_csv('3. Police Data.csv')
```

```python
print("--- 数据预览 ---")
print(df.head())
print("\n--- 数据基本信息 ---")
print(df.info())
```

```python
print("\n--- 缺失值统计 ---")
print(df.isnull().sum())

cols_to_drop = [col for col in df.columns if df[col].isnull().sum() == len(df)]

df.drop(columns=cols_to_drop, inplace=True)
print(f"\n已删除全空列: {cols_to_drop}")
```

```python
speeding_df = df[df['violation'] == 'Speeding']

gender_counts = speeding_df['driver_gender'].value_counts()
print("\n--- Q3-2: 超速拦截中的性别分布 ---")
print(gender_counts)
```

```python
search_by_gender = df.groupby('driver_gender')['search_conducted'].mean()
print("\n--- Q3-3: 不同性别被搜查的比例 ---")
print(search_by_gender)
```

```python
print("\n--- stop_duration 的唯一值 ---")
print(df['stop_duration'].unique())

duration_mapping = {
    '0-15 Min': 7.5,
    '16-30 Min': 23,
    '30+ Min': 45
}

df['stop_duration_numeric'] = df['stop_duration'].map(duration_mapping)

mean_duration = df['stop_duration_numeric'].mean()
print(f"\n--- Q3-4: 平均拦截时长 (分钟) ---")
print(mean_duration)
```

```python
age_distribution = df.groupby('violation')['driver_age'].describe()
print("\n--- Q3-5: 不同违规类型的年龄分布 ---")
print(age_distribution)
```
